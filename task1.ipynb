```python
m, n = X_train.shape
w = np.zeros(n)
b = 0.0
lr = 0.1
epochs = 100

def r2(y, y_hat):
    return 1 - np.sum((y - y_hat)**2) / np.sum((y - y.mean())**2)

costs, train_acc, test_acc = [], [], []

for epoch in range(epochs):
    y_hat = X_train @ w + b          # 1. predict
    err = y_hat - y_train            # 2. error
    cost = np.sum(err**2) / (2*m)    # cost
    dw = (X_train.T @ err) / m       # 3. gradients
    db = np.mean(err)
    w = w - lr * dw                  # 4. update
    b = b - lr * db

    costs.append(cost)
    train_acc.append(r2(y_train, X_train @ w + b))
    test_acc.append(r2(y_test, X_test @ w + b))
    print(f"Epoch {epoch+1}: cost={cost:.4f}, train R2={train_acc[-1]:.4f}, test R2={test_acc[-1]:.4f}")
```

```python
import matplotlib.pyplot as plt
plt.plot(costs)
plt.xlabel("Epoch")
plt.ylabel("Cost")
plt.title("Cost vs Epoch")
plt.show()
           
```
